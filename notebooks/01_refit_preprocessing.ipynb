```python
df_hourly = df.resample('h').mean()
```

```python
df_hourly['hour'] = df_hourly.index.hour
df_hourly['weekday'] = df_hourly.index.weekday
df_hourly['month'] = df_hourly.index.month
```

```python
train = df_hourly.loc['2013-10-09':'2014-12-31']
test  = df_hourly.loc['2015-01-01':'2015-06-30']
```

```python
from sklearn.linear_model import LinearRegression

X_train = train[['hour','weekday']]
y_train = train['Aggregate']

model = LinearRegression()
model.fit(X_train, y_train)

print("Baseline model trained.")
```

```python
# Add new lag features
df_hourly['lag48'] = df_hourly['Aggregate'].shift(48)
df_hourly['lag168'] = df_hourly['Aggregate'].shift(168)

# Train/test split
train = df_hourly.loc['2013-10-09':'2014-12-31'].copy()
test  = df_hourly.loc['2015-01-01':'2015-06-30'].copy()

from sklearn.ensemble import RandomForestRegressor
import pandas as pd

# Features including new lags
features = ['hour','weekday','month','lag1','lag24','rolling3','lag48','lag168']

X_train = train[features].dropna()
y_train = train['Aggregate'].loc[X_train.index]

model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Feature importance
importance = model.feature_importances_
pd.DataFrame({'feature':features, 'importance':importance}).sort_values('importance', ascending=False)
```

```python
X_test = test[['hour','weekday','month','lag1','lag24','rolling3','lag48','lag168']].dropna()
y_test = test['Aggregate'].loc[X_test.index]

y_pred = model.predict(X_test)

from sklearn.metrics import mean_squared_error, r2_score
print("MSE:", mean_squared_error(y_test, y_pred))
print("R²:", r2_score(y_test, y_pred))
```

```python
plt.figure(figsize=(12,6))
plt.plot(y_test.index[:1000], y_test[:1000], label="Actual")
plt.plot(y_test.index[:1000], y_pred[:1000], label="Predicted")
plt.legend()
plt.show()
```
